```python
principle = int(input("Enter principal: "))
rate = int(input("Enter rate: "))
time = int(input("Enter time: "))

def simple_interest(principle, rate, time):
    return (principle * rate * time) / 100

si = simple_interest(principle, rate, time)

print("Simple Interest =", si)
```

```python
#default argument
def msg1(name , age=22):
  print(name)
  print(age)
```

```python
#variable length
def msg2(arg1,*arg2):
  print(arg1)
  for x in arg2:
    print(x)
```

```python
msg2(1,2,3,4,5)
```
